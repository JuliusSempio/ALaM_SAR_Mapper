# ASTI-ALaM Implementation for DTW-Based Crop Mapping
## Using DTAIDistance module from KU Leuven
https://dtaidistance.readthedocs.io/en/latest/usage/dtw.html

## Version 2.0 β1

### Features
- Purely focused UI for end users
- All back-end codes run in their respective modules

### Requirements
- pre-install the Python modules in the requirements.txt file
- all vector shapefiles (at least one reference parcel and one study area boundary) must be loaded in the shp folder
- all images should be in GeoTIFF format and stored in their designated time-series folders in the img_data folder

### Anticipated TODOs for Future Coders
- better handling of filename conventions used by different satellite platforms
- use start and end dates to limit the number of images needed to be accounted for in a directory
- implement a lot of necessary exception handling

### Known Issues
- Misconfigurations will lead to red letter errors due to absence of proper exception handling (i.e. unmatched dates)
- Current date extraction code from filenames is inflexible - need for it to be flexible ("generate_temporal_signature module" lines 22 - 28)
- The "use_pruning" parameter in DTAIdistance is not working as intended in newer versions, so is temporarily disabled in the "generate_accumulated_distance_cost_map" module (line 66)
